# Gold through war and inflation

Empirical note, synthetic seeded data. Runs offline.


```python
import numpy as np, pandas as pd
rng = np.random.default_rng(99)
```

## Synthetic gold and 10y real yield


```python
m = pd.date_range('2005-01-31', periods=240, freq='ME')
real = pd.Series(1.0+np.cumsum(rng.normal(0,0.15,len(m))), index=m).clip(-1.5,4)
gold_ret = -0.9*real.diff().fillna(0) + rng.normal(0,0.03,len(m))
gold = 100*np.exp(np.cumsum(gold_ret))
```

## Gold returns vs change in real yield


```python
df = pd.DataFrame({'gold_ret':gold_ret,'dreal':real.diff()}).dropna()
print('beta', round(np.polyfit(df['dreal'],df['gold_ret'],1)[0],2), 'corr', round(df.corr().iloc[0,1],2))
```

## Returns by inflation regime


```python
infl = pd.Series(rng.normal(2.5,1.5,len(m)), index=m)
hi = infl>infl.median(); fall = real.diff()<0
g = pd.Series(gold_ret.values, index=m)
print('hi&falling', round(g[hi&fall].mean(),4),'hi&rising', round(g[hi&~fall].mean(),4),'low', round(g[~hi].mean(),4))
```
