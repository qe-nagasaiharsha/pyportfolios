# The Kelly criterion for position sizing

Runnable companion to the tutorial. Synthetic, seeded — runs offline.


```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(7)
```

## The two forms


```python
def kelly_discrete(p, b):
    q = 1 - p
    return (p*b - q) / b

def kelly_continuous(mu, sigma):
    return mu / sigma**2

print(kelly_discrete(0.55, 1.0), kelly_continuous(0.08, 0.16))
```

## Growth curve g(f)


```python
p, b = 0.55, 1.0
f = np.linspace(0, 0.6, 200)
g = p*np.log(1+b*f) + (1-p)*np.log(1-f)
plt.plot(f, g); plt.axvline(kelly_discrete(p,b), ls='--'); plt.xlabel('f'); plt.ylabel('g(f)'); plt.show()
```

## Full vs half Kelly wealth paths


```python
def simulate(frac, p=0.55, b=1.0, n=500, paths=400):
    wins = rng.random((paths, n)) < p
    step = np.where(wins, 1 + b*frac, 1 - frac)
    return np.cumprod(step, axis=1)

full = simulate(kelly_discrete(p,b)); half = simulate(0.5*kelly_discrete(p,b))
print('full median', np.median(full[:,-1]).round(2), '5pct', np.percentile(full[:,-1],5).round(2))
print('half median', np.median(half[:,-1]).round(2), '5pct', np.percentile(half[:,-1],5).round(2))
```
